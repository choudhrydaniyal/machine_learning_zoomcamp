```python
u = np.array([1, 0.5, 2])
v = np.array([5, 6, 7])
w = np.array([[2, 4, 5],
     [1, 2, 1],
     [3, 1, 2]])
x = np.array([[3, 6, 1, 2],
             [1, 0.5, 2, 1],
             [4, 2, 1.5, 0.5]])
```

```python
print("u: ", u.shape)
print("v: ", v.shape)
print("w: ", w.shape)
print("x: ", x.shape)
```

```python
def vector_vector_multiplication(u, v):
    assert u.shape[0] == v.shape[0]

    n = u.shape[0]

    result = 0.0

    for i in range(n):
        result = result + u[i] * v[i]

    return result
```

```python
print(vector_vector_multiplication(u, v))
```

```python
u.dot(v)
```

```python
def matrix_vector_multiplication(u, v):
    assert u.shape[1] == v.shape[0]

    no_of_rows = u.shape[0]
    

    result = np.zeros(no_of_rows)

    for i in range(no_of_rows):
        result[i] = vector_vector_multiplication(u[i], v)

    return result
```

```python
print(matrix_vector_multiplication(w, u))
```

```python
w.dot(u)
```

```python
def matrix_matrix_multiplication(u, v):
    assert u.shape[1] == v.shape[0]

    no_of_rows = u.shape[0]
    no_of_cols = v.shape[1]

    result = np.zeros((no_of_rows, no_of_cols))

    for i in range(no_of_cols):
        vi = v[:, i]
        result[:, i] = matrix_vector_multiplication(u, vi)

    return result
```

```python
print(matrix_matrix_multiplication(w, x))
```

```python
w.dot(x)
```
